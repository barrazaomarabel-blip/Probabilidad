```python

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
x = [0, 1, 2, 3, 4, 5]
p = [0.10, 0.25, 0.30, 0.20, 0.10, 0.05]
suma_probabilidades = 0
for i in range(len(p)):
  suma_probabilidades += p[i]
print(f"Suma total de probabilidades: {suma_probabilidades:.2f}")
if round(suma_probabilidades, 2) == 1.0:
  print("-> Correcto: Las probabilidades suman 1 exactamente.\n")
else:
  print("-> Advertencia: Las probabilidades no suman 1.\n")
tabla_distribucion = pd.DataFrame({"Número de células anormales (X)": x, "Probabilidad P(X)": p})
display(tabla_distribucion)
print("\n")
valor_esperado = 0
for i in range(len(x)):
  xi = x[i]
  pi = p[i]
  producto = xi * pi
  valor_esperado += producto
  print(f"x = {xi}, p = {pi}, x * p = {producto:.2f}")
print(f"\nValor esperado (Esperanza) E(X): {valor_esperado:.2f} células"" anormales por muestra\n")
e_x2 = 0
for i in range(len(x)):
  xi = x[i]
  pi = p[i]
  prod_cuadrado = (xi**2) * pi
  e_x2 += prod_cuadrado
varianza = e_x2 - (valor_esperado**2)
desviacion_estandar = np.sqrt(varianza)
print(f"E(X^2) = {e_x2:.2f}")
print(f"Varianza Var(X): {varianza:.4f}")
print(f"Desviación Estándar: {desviacion_estandar:.4f}\n")
f_acumulada = []
acumulador = 0
for i in range(len(p)):
  acumulador += p[i]
  f_acumulada.append(acumulador)
tabla_acumulada = pd.DataFrame({"Células (x)": x, "P(X = x)": p, "F(x) [Acumulada]": f_acumulada})
display(tabla_acumulada)
print("\n")
plt.figure(figsize=(14, 5))
plt.subplot(1, 2, 1)
plt.bar([str(val) for val in x], p, color="cornflowerblue", edgecolor="black", width=0.5)
plt.title("Función de Probabilidad P(X)")
plt.xlabel("Número de células anormales (X)")
plt.ylabel("P(X)")
plt.ylim(0, 0.4)
for i, val in enumerate(p):
  plt.text(i, val + 0.015, f"{val:.2f}", ha="center", fontweight="bold")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.subplot(1, 2, 2)
plt.step(x,f_acumulada,where="post",color="darkorange",linewidth=2,label="F(x)",)
plt.plot(x, f_acumulada, "ro", label="Puntos acumulados")
plt.title("Función de Distribución Acumulada F(x)")
plt.xlabel("Células (x)")
plt.ylabel("F(x)")
plt.ylim(0, 1.1)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
print(f"• ¿Cuál es el promedio esperado de células anormales por muestra?\n"f"  Se espera encontrar en promedio {valor_esperado:.2f} células anormales"f" por muestra analizada (E(X)).\n")
print(f"• ¿Existe mucha variabilidad entre las muestras?\n"f"  Con una desviación estándar de {desviacion_estandar:.2f} células, se"f" observa una dispersión moderada alrededor del promedio ({round(valor_esperado)}"" células), indicando el grado de heterogeneidad en el tejido de los"" pacientes.\n")
print( f"• ¿Qué indica la función acumulada sobre la probabilidad de encontrar" f" un número bajo o alto de células anormales?\n"f"  Indica que hay una probabilidad acumulada del"f" {f_acumulada[2]*100:.1f}\n  de encontrar 2 células o menos en una muestra" f" seleccionada al azar, mientras que la probabilidad de hallar un conteo"f" alto (4 o 5 células) es considerablemente menor.")
```

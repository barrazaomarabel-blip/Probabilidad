```python

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
x = [0, 10, 20, 50]
p = [0.50, 0.30, 0.15, 0.05]
suma_probabilidades = 0
for i in range(len(p)):
  suma_probabilidades += p[i]
print(f"Suma total de probabilidades: {suma_probabilidades:.2f}")
if round(suma_probabilidades, 2) == 1.0:
  print("-> Correcto: Las probabilidades suman 1.\n")
else:
  print("-> Advertencia: Las probabilidades no suman 1.\n")
tabla_distribucion = pd.DataFrame( {"Premio (X)": ["A", "B", "C", "D"],"Monto (pesos)": x,"Probabilidad P(X)": p, })
display(tabla_distribucion)
print("\n")
valor_esperado = 0
for i in range(len(x)):
  xi = x[i]
  pi = p[i]
  producto = xi * pi
  valor_esperado += producto
  print(f"x = {xi}, p = {pi}, x * p = {producto:.2f}")
print(f"\nValor esperado (Esperanza) E(X): {valor_esperado:.2f} pesos\n")
e_x2 = 0
for i in range(len(x)):
  xi = x[i]
  pi = p[i]
  prod_cuadrado = (xi**2) * pi
  e_x2 += prod_cuadrado
varianza = e_x2 - (valor_esperado**2)
desviacion_estandar = np.sqrt(varianza)
print(f"E(X^2) = {e_x2:.2f}")
print(f"Varianza Var(X): {varianza:.4f}")
print(f"Desviación Estándar: {desviacion_estandar:.4f}\n")
f_acumulada = []
acumulador = 0
for i in range(len(p)):
  acumulador += p[i]
  f_acumulada.append(acumulador)
tabla_acumulada = pd.DataFrame({"Monto x": x, "P(X = x)": p, "F(x) [Acumulada]": f_acumulada})
display(tabla_acumulada)
print("\n")
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.bar([str(val) for val in x], p, color="skyblue", edgecolor="black", width=0.4)
plt.title("Función de Probabilidad P(X)")
plt.xlabel("Monto en pesos (X)")
plt.ylabel("P(X)")
plt.ylim(0, 0.6)
for i, val in enumerate(p):
  plt.text(i, val + 0.02, f"{val:.2f}", ha="center", fontweight="bold")
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.subplot(1, 2, 2)
plt.step(x,f_acumulada,where="post",color="darkorange",linewidth=2,label="F(x)",)
plt.plot(x, f_acumulada, "ro", label="Puntos")
plt.title("Función de Distribución Acumulada F(x)")
plt.xlabel("Monto en pesos (x)")
plt.ylabel("F(x)")
plt.ylim(0, 1.1)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()
plt.tight_layout()
plt.show()
print(f"• ¿Qué significa el valor esperado? Significa que, a largo plazo, el"f" jugador puede esperar ganar en promedio ${valor_esperado:.2f} pesos por"" cada boleto de la tómbola.")
print("• ¿El juego es favorable? Dependerá estrictamente del costo de adquisición"f" del boleto: si cuesta menos de ${valor_esperado:.2f} pesos, el juego es" f" favorable; si cuesta más, no le conviene al jugador.")
```

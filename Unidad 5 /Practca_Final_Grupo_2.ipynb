```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, linregress
from sklearn.linear_model import LinearRegression
from google.colab import files
```

```python
archivos = files.upload()
```

```python
grupo2 = pd.read_csv("grupo2.csv")
```

```python
print(grupo2.head())
```

```python
print(grupo2.describe())
```

```python
plt.hist(grupo2["Calificacion"], bins=10, edgecolor="black")
plt.title("Histograma de calificaciones - Grupo 2")
plt.xlabel("Calificacion")
plt.ylabel("Frecuencia")
plt.show()
```

```python
X = grupo2[["HorasEstudio"]]
Y = grupo2["Calificacion"]
```

```python
modelo2 = LinearRegression()
modelo2.fit(X, Y)
```

```python
pendiente2 = modelo2.coef_[0]
intercepto2 = modelo2.intercept_
```

```python
print("Ecuacion: Y =", pendiente2, "X +", intercepto2)
```

```python
pearson2, p2 = pearsonr(grupo2["HorasEstudio"], grupo2["Calificacion"])
print("Coeficiente de Pearson:", pearson2)
```

```python
r2_2 = modelo2.score(X, Y)
print("R cuadrada:", r2_2)
```

```python
orden2 = np.argsort(grupo2["HorasEstudio"].values)
plt.scatter(grupo2["HorasEstudio"], grupo2["Calificacion"])
plt.plot(grupo2["HorasEstudio"].values[orden2], modelo2.predict(X)[orden2], color="red")
plt.title("Regresion lineal - Grupo 2")
plt.xlabel("Horas de estudio")
plt.ylabel("Calificacion")
plt.show()
```

```python
print("Por cada hora adicional de estudio, la calificacion cambia", pendiente2, "puntos.")
```

```python
if abs(pearson2) < 0.30:
    print("La correlacion es muy debil.")
elif abs(pearson2) < 0.60:
    print("La correlacion es moderada.")
elif abs(pearson2) < 0.80:
    print("La correlacion es fuerte.")
else:
    print("La correlacion es muy fuerte.")
    if pearson2 > 0:
        print("La relacion es positiva.")
    elif pearson2 < 0:
        print("La relacion es negativa.")
    else:
        print("No existe relacion lineal.")
```

```python
print("Valor p:", p2)
if p2 < 0.05:
    print("La relacion es estadisticamente significativa.")
else:
    print("La relacion no es estadisticamente significativa.")
```

```python
print("El modelo explica", r2_2 * 100, "% de la variacion de las calificaciones.")
if r2_2 < 0.30:
    print("Las predicciones del modelo tienen baja confiabilidad.")
elif r2_2 < 0.70:
    print("Las predicciones tienen confiabilidad moderada.")
else:
    print("Las predicciones tienen buena confiabilidad.")
```

```python
print("Conclusion: En el Grupo 2 existe una relacion positiva fuerte entre las horas de estudio y las calificaciones.")
```

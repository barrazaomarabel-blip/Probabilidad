```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, linregress
from sklearn.linear_model import LinearRegression
from google.colab import files
```

```python
archivos = files.upload()
```

```python
grupo3 = pd.read_csv("grupo3.csv")
```

```python
print(grupo3.head())
```

```python
print(grupo3.describe())
```

```python
plt.hist(grupo3["Calificacion"], bins=10, edgecolor="black")
plt.title("Histograma de calificaciones - Grupo 3")
plt.xlabel("Calificacion")
plt.ylabel("Frecuencia")
plt.show()
```

```python
X = grupo3[["HorasEstudio"]]
Y = grupo3["Calificacion"]
```

```python
modelo3 = LinearRegression()
modelo3.fit(X, Y)
```

```python
pendiente3 = modelo3.coef_[0]
intercepto3 = modelo3.intercept_
```

```python
print("Ecuacion: Y =", pendiente3, "X +", intercepto3)
```

```python
pearson3, p3 = pearsonr(grupo3["HorasEstudio"], grupo3["Calificacion"])
print("Coeficiente de Pearson:", pearson3)
```

```python
r2_3 = modelo3.score(X, Y)
print("R cuadrada:", r2_3)
```

```python
orden3 = np.argsort(grupo3["HorasEstudio"].values)
plt.scatter(grupo3["HorasEstudio"], grupo3["Calificacion"])
plt.plot(grupo3["HorasEstudio"].values[orden3], modelo3.predict(X)[orden3], color="red")
plt.title("Regresion lineal - Grupo 3")
plt.xlabel("Horas de estudio")
plt.ylabel("Calificacion")
plt.show()
```

```python
print("Por cada hora adicional de estudio, la calificacion cambia", pendiente3, "puntos.")
```

```python
if abs(pearson3) < 0.30:
    print("La correlacion es muy debil.")
elif abs(pearson3) < 0.60:
    print("La correlacion es moderada.")
elif abs(pearson3) < 0.80:
    print("La correlacion es fuerte.")
else:
    print("La correlacion es muy fuerte.")
    if pearson3 > 0:
        print("La relacion es positiva.")
    elif pearson3 < 0:
        print("La relacion es negativa.")
    else:
        print("No existe relacion lineal.")
```

```python
print("Valor p:", p3)
if p3 < 0.05:
    print("La relacion es estadisticamente significativa.")
else:
    print("La relacion no es estadisticamente significativa.")
```

```python
print("El modelo explica", r2_3 * 100, "% de la variacion de las calificaciones.")
if r2_3 < 0.30:
    print("Las predicciones del modelo tienen baja confiabilidad.")
elif r2_3 < 0.70:
    print("Las predicciones tienen confiabilidad moderada.")
else:
    print("Las predicciones tienen buena confiabilidad.")
```

```python
print("Conclusion: En el Grupo 3 existe una relacion positiva muy fuerte entre las horas de estudio y las calificaciones.")
```

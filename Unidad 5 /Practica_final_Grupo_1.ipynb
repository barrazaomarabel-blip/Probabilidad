```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, linregress
from sklearn.linear_model import LinearRegression
from google.colab import files
```

```python
archivos = files.upload()
```

```python
grupo1 = pd.read_csv("grupo1.csv")
```

```python
print(grupo1.head())
```

```python
print(grupo1.describe())
```

```python
plt.hist(grupo1["Calificacion"], bins=10, edgecolor="black")
plt.title("Histograma de calificaciones - Grupo 1")
plt.xlabel("Calificacion")
plt.ylabel("Frecuencia")
plt.show()
```

```python
X = grupo1[["HorasEstudio"]]
Y = grupo1["Calificacion"]
```

```python
modelo1 = LinearRegression()
modelo1.fit(X, Y)
```

```python
pendiente1 = modelo1.coef_[0]
intercepto1 = modelo1.intercept_
```

```python
print("Ecuacion: Y =", pendiente1, "X +", intercepto1)
```

```python
pearson1, p1 = pearsonr(grupo1["HorasEstudio"], grupo1["Calificacion"])
print("Coeficiente de Pearson:", pearson1)
```

```python
r2_1 = modelo1.score(X, Y)
print("R cuadrada:", r2_1)
```

```python
orden1 = np.argsort(grupo1["HorasEstudio"].values)
plt.scatter(grupo1["HorasEstudio"], grupo1["Calificacion"])
plt.plot(grupo1["HorasEstudio"].values[orden1], modelo1.predict(X)[orden1], color="red")
plt.title("Regresion lineal - Grupo 1")
plt.xlabel("Horas de estudio")
plt.ylabel("Calificacion")
plt.show()
```

```python
print("Por cada hora adicional de estudio, la calificacion cambia", pendiente1, "puntos.")
```

```python
if abs(pearson1) < 0.30:
    print("La correlacion es muy debil.")
elif abs(pearson1) < 0.60:
    print("La correlacion es moderada.")
elif abs(pearson1) < 0.80:
    print("La correlacion es fuerte.")
else:
    print("La correlacion es muy fuerte.")
    if pearson1 > 0:
        print("La relacion es positiva.")
    elif pearson1 < 0:
        print("La relacion es negativa.")
    else:
        print("No existe relacion lineal.")
```

```python
print("Valor p:", p1)
if p1 < 0.05:
    print("La relacion es estadisticamente significativa.")
else:
    print("La relacion no es estadisticamente significativa.")
```

```python
print("El modelo explica", r2_1 * 100, "% de la variacion de las calificaciones.")
if r2_1 < 0.30:
    print("Las predicciones del modelo tienen baja confiabilidad.")
elif r2_1 < 0.70:
    print("Las predicciones tienen confiabilidad moderada.")
else:
    print("Las predicciones tienen buena confiabilidad.")
```

```python
print("Conclusion: En el Grupo 1 existe una relacion lineal practicamente nula entre las horas de estudio y las calificaciones.")
```

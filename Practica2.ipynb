```python
import random
import matplotlib.pyplot as plt # Corrected typo here
from statistics import *
```

```python
import random

## Generamos 100 numeros aleatorios que representan la produccion
# en metros de tela sw 100 dias
fabrica1 = []
for i in range(100):
  numero_aleatorio1 = round(random.uniform(50, 100), 2)
  fabrica1.append(numero_aleatorio1)
  print(fabrica1)
```

```python
fabrica2 = []
for i in range(100):
  numero_aleatorio2 = round(random.uniform(70, 100), 2)
  fabrica2.append(numero_aleatorio2)
  print(fabrica2)
```

```python
maximo1 = max(fabrica1)
minimo1 = min(fabrica1)
rango1 = maximo1 - minimo1
media1 = mean(fabrica1)
mediana1 = median(fabrica1)
moda1 = mode(fabrica1)
desviacion_estandar1 = stdev(fabrica1)
print( "Estadistica descriptiva para la fabrica 1 \n" "Media =", media1,
      "\n Mediana =" mediana1, "\n Moda =", moda1, "\n Maximo =", maximo1,
       "\n Minimo =", minimo1, "\n Rango =", rango1)
print(desviacion_estandar_fabrica1)
```

```python
maximo2 = max(fabrica2)
minimo2 = min(fabrica2)
rango2 = maximo2 - minimo2
media2 = mean(fabrica2)
mediana2 = median(fabrica2)
moda2 = mode(fabrica2)
desviacion_estandar2 = stdev(fabrica2) # Corrected variable name
print( "Estadistica descriptiva para la fabrica 2 \n" "Media =", media2,
      "\n Mediana =", mediana2, "\n Moda =", moda2, "\n Maximo =", maximo2,
       "\n Minimo =", minimo2, "\n Rango =", rango2)
print("Desviación estándar =", desviacion_estandar2) # Corrected variable name and added descriptive text
```

```python
plt.figure(figsize = (8,6) )
plt.hist(fabrica1, bins=10, alpha=0.2, color='b', edgecolor='black')
plt.axvline(media1, color'b', linestyles='dashed', linewidth=1, label=f'Media = {media1: 2f}')
plt.axvline(mediana1, color'r', linestyles='dashed', linewidth=1, label=f'Mediana = {mediana1: 2f}')
plt.axvline(moda1, color'y', linestyles='dashed', linewidth=1, label=f'Moda = {moda1: 2f}')
plt.axvline(media1 - desviacion_estandar_fabrica1, color='black', linestyles='dashed',
            inewidth=2, label=f'Desv. Estandar = {desviacion_estandar_fabrica1: 2f}')
plt.axvline(media1 + desviacion_estandar_fabrica1, color='black', linestyles='dashed',
            inewidth=2,)
plt.xlabel('valor')
plt.ylabel('frecuenca')
plt.title("produccion diaria de la fabrica 1")
plt.legend()
plt.show()

```

```python
plt.figure(figsize = (8,6) )
plt.hist(fabrica2, bins=10, alpha=0.2, color='b', edgecolor='black')
plt.axvline(media2, color='b', linestyles='dashed', linewidth=1, label=f'Media = {media2: 2f}')
plt.axvline(mediana2, color='r', linestyles='dashed', linewidth=1, label=f'Mediana = {mediana2: 2f}')
plt.axvline(moda2, color='y', linestyles='dashed', linewidth=1, label=f'Moda = {moda2: 2f}')
plt.axvline(media2 - desviacion_estandar2, color='black', linestyles='dashed',
            linewidth=2, label=f'Desv. Estandar = {desviacion_estandar2: 2f}') # Corrected variable name and typo
plt.axvline(media2 + desviacion_estandar2, color='black', linestyles='dashed',
            linewidth=2,) # Corrected variable name and typo
plt.xlabel('valor')
plt.ylabel('frecuenca')
plt.title("produccion diaria de la fabrica 2") # Corrected title
plt.legend()
plt.show()
```

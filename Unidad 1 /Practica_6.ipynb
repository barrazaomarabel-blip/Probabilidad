```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import math
```

```python
# Cargar el archivo CSV
autos = pd.read_csv("autos2 (1).csv")
# Mostrar los primeros registros
autos.head()
```

```python
print("Número de registros:", autos.shape[0])
print("Número de columnas:", autos.shape[1])
```

```python
# Eliminar valores vacíos de la columna price
precios = autos['price'].dropna()
print("Cantidad de precios disponibles:", len(precios))
```

```python
# Determinar el valor mínimo de precios
precio_minimo = precios.min()
print("El precio mínimo de los automóviles es:", precio_minimo)
```

```python
# Determinar el valor máximo de precios
precio_maximo = precios.max()
print("El precio máximo de los automóviles es:", precio_maximo)
```

```python
# Determinar la moda de los precios
moda_precios = precios.mode()
print("La moda de los precios es:")
print(moda_precios)
```

```python
# Determinar la media de los precios
media_precios = precios.mean()
print("La media de los precios es:", media_precios)
```

```python
# Determinar la mediana de los precios
mediana_precios = precios.median()
print("La mediana de los precios es:", mediana_precios)
```

```python
# Determinar la desviación estándar de los precios
desviacion = precios.std()
print("La desviación estándar de los precios es:", desviacion)
```

```python
# Mostrar los estadísticos y cuartiles
autos['price'].describe()
```

```python
print("ESTADÍSTICAS DE PRECIOS")
print("Precio mínimo:", precio_minimo)
print("Precio máximo:", precio_maximo)
print("Moda:")
print(moda_precios.tolist())
print("Media:", round(media_precios, 2))
print("Mediana:", mediana_precios)
print("Desviación estándar:", round(desviacion, 2))
print("\nESTADÍSTICOS Y CUARTILES")
print(autos['price'].describe())
```

```python
# Cantidad de datos
n = len(precios)
# Regla de Sturges
bins_sturges = math.ceil(1 + math.log2(n))
print("Cantidad de datos:", n)
print("Número óptimo de bins según Sturges:", bins_sturges)
```

```python
# Histograma de precios
plt.figure(figsize=(10, 6))
plt.hist(precios,bins=bins_sturges,edgecolor='black',alpha=0.7)
plt.title("Histograma de Precios de los Automóviles")
plt.xlabel("Precio")
plt.ylabel("Frecuencia")
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='body-style',y='price',data=autos)
plt.title("Distribución de Precios por Tipo de Vehículo")
plt.xlabel("Tipo de Vehículo")
plt.ylabel("Precio")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='drive-wheels',y='price',data=autos)
plt.title("Distribución de Precios por Tipo de Tracción")
plt.xlabel("Tipo de Tracción")
plt.ylabel("Precio")
plt.show()
```

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='engine-location',y='price',data=autos)
plt.title("Distribución de Precios por Ubicación del Motor")
plt.xlabel("Ubicación del Motor")
plt.ylabel("Precio")
plt.show()
```

```python
print("PRECIO POR TIPO DE VEHÍCULO")
print(autos.groupby('body-style')['price'].agg(['count', 'mean', 'median', 'min', 'max']))
print("\nPRECIO POR TIPO DE TRACCIÓN")
print(autos.groupby('drive-wheels')['price'].agg(['count', 'mean', 'median', 'min', 'max']))
print("\nPRECIO POR UBICACIÓN DEL MOTOR")
print(autos.groupby('engine-location')['price'].agg(['count', 'mean', 'median', 'min', 'max']))
```
